```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Dataset Generate Karein (Direct Code Se Data Ban Jayega)
np.random.seed(42)
n = 500

df = pd.DataFrame({
    'order_weight_kg': np.round(np.random.gamma(shape=2.0, scale=3.5, size=n) + 0.5, 2),
    'cycle_time_hrs': np.round(np.random.normal(loc=45, scale=15, size=n).clip(12, 120), 1),
})
df['freight_cost_inr'] = np.round(150 + (df['order_weight_kg'] * 35) + (df['cycle_time_hrs'] * 2.5) + np.random.normal(0, 25, n), 2)

# 2. Correlation Matrix Calculate Karein
corr_matrix = df[['order_weight_kg', 'cycle_time_hrs', 'freight_cost_inr']].corr()

# 3. Correlation Heatmap Plot Karein
sns.set_theme(style="whitegrid")
plt.figure(figsize=(6, 4.5))
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=1.5)
plt.title('Figure 4: Correlation Matrix Heatmap')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Dataset Generate Karein
np.random.seed(42)
n = 500

df = pd.DataFrame({
    'order_weight_kg': np.round(np.random.gamma(shape=2.0, scale=3.5, size=n) + 0.5, 2),
    'cycle_time_hrs': np.round(np.random.normal(loc=45, scale=15, size=n).clip(12, 120), 1),
})
df['freight_cost_inr'] = np.round(150 + (df['order_weight_kg'] * 35) + (df['cycle_time_hrs'] * 2.5) + np.random.normal(0, 25, n), 2)

# 2. Scatter Plot + Regression Line Plot Karein
sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 4.5))
sns.regplot(x='order_weight_kg', y='freight_cost_inr', data=df,
            scatter_kws={'alpha':0.5, 'color':'#2B6CB0'},
            line_kws={'color':'#C53030', 'linewidth':2})

plt.title('Figure 3: Linear Relationship - Order Weight vs. Freight Cost', fontsize=11, fontweight='bold', pad=12)
plt.xlabel('Package Mass (Kg)', fontsize=10, fontweight='bold')
plt.ylabel('Freight Shipping Fee (INR)', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Dataset Generate Karein
np.random.seed(42)
n = 500
carriers = ['ExpressLog', 'BlueDart_3PL', 'SurfaceFast', 'Delhivery_Express']

df = pd.DataFrame({
    'carrier_partner': np.random.choice(carriers, size=n, p=[0.3, 0.25, 0.25, 0.2]),
    'cycle_time_hrs': np.round(np.random.normal(loc=45, scale=15, size=n).clip(12, 120), 1),
})
df.loc[df['carrier_partner'] == 'SurfaceFast', 'cycle_time_hrs'] += np.random.exponential(scale=20, size=(df['carrier_partner'] == 'SurfaceFast').sum())

# Boxplot Plot Karein
sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 4.5))
sns.boxplot(x='carrier_partner', y='cycle_time_hrs', data=df, palette='Blues_r')

plt.title('Figure 1: Cycle Time Distribution Across 3PL Carrier Partners', fontsize=11, fontweight='bold', pad=12)
plt.xlabel('Carrier Partner', fontsize=10, fontweight='bold')
plt.ylabel('Order Cycle Time (Hours)', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Dataset Generate Karein
np.random.seed(42)
n = 500

df = pd.DataFrame({
    'order_weight_kg': np.round(np.random.gamma(shape=2.0, scale=3.5, size=n) + 0.5, 2),
    'cycle_time_hrs': np.round(np.random.normal(loc=45, scale=15, size=n).clip(12, 120), 1),
})
df['freight_cost_inr'] = np.round(150 + (df['order_weight_kg'] * 35) + (df['cycle_time_hrs'] * 2.5) + np.random.normal(0, 25, n), 2)

# Histogram + KDE Plot Karein
sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 4.5))
sns.histplot(df['freight_cost_inr'], kde=True, color='#1A365D', bins=25)

plt.title('Figure 2: Freight Cost Distribution & Density (INR)', fontsize=11, fontweight='bold', pad=12)
plt.xlabel('Freight Cost (INR)', fontsize=10, fontweight='bold')
plt.ylabel('Order Frequency', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.show()
```
